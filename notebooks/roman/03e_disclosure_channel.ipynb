29 of the 31 targets are first stated in the prepared remarks, regardless of whether the reports carry the figure (A), mention the commitment without it (B) or not at all (C). ## Part 2: the call is earlier than the reports

```python
cf = pd.read_csv(OUT / "call_first.csv", keep_default_na=False)
cf[["unit", "call_date", "first_report", "report_date", "lead_months", "strength", "call_quote", "call_location",
    "report_quote", "report_location"]]
```

Five UBS targets appear in a report only after the call, 1.7 to 9.2 months later (publication dates are the signature dates of the filings; each quarterly report is signed on the day of its call). Roman confirmed two cases: the CET1 guidance of around 14 % (call 31 August 2023, report 6 February 2024, 5.2 months) and the effective tax rate of around 40 % (call 6 February 2024, report 7 May 2024, 3.0 months).

The other three are weak: the Basel III final estimate of about 5 % (9.2 months) had an equivalent figure of USD 25bn in the 2023-Q4 report, the funding-cost savings (1.7 months) and the NCL share below 5 % (4.5 months) reach the annual report, which follows the fourth-quarter call by construction, and the NCL case differs from earlier reports only in wording (“below” against “around 5%”).
