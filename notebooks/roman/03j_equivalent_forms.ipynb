## Why

T32 was classed as mentioned without a figure, although the reports give the same increase in another unit (around USD 10bn of tier 1 capital against around 180 basis points). This phase searches the remaining eight units and T30 in every equivalent form of the call figure.

## Equivalent forms

```python
forms = pd.read_csv(OUT / "forms.csv")
forms["searched values"] = forms.terms.str.split(" | ", regex=False).str.len()
forms[["unit", "form", "calculation", "call_figure", "central_value", "band_low", "band_high", "searched values"]]
```

Each form is a function of the call figure; its band is the image of the call figure plus and minus 10%. T68 is searched in two readings (`T68_r1`: gone concern capital, `T68_r2`: all funding raised through the holding company).

## Search

```python
log = json.loads((OUT / "run_log.json").read_text())
pd.Series(log).to_frame("value")
```

```python
summary = pd.read_csv(OUT / "summary.csv")
summary[["unit", "forms", "hits_pass_A", "hits_pass_B", "hits_pass_B_read", "passages_read", "distinct_passages_read"]]
```

Pass A (a figure form near a keyword) gives 904 hits, all read. Pass B (keywords alone) gives 11,233 hits, of which the 1,332 with a forward-looking word and a quantity in the context were read. Overlapping contexts form 874 passages; no unit exceeds the cap of 600.

## Classes of the passages read

```python
classes = ["same target, same unit", "same target, another unit", "same target, another figure", "result", "definition or requirement", "other"]
summary[["unit"] + classes]
```

No passage states a call figure in the unit of the call, and none states another figure for the same target. One passage, for T30, gives the target in an equivalent form. The split of the remaining passages into result, definition or requirement and other is assigned by a wording rule and only sorts them; none of them states a target for its unit.

## The one equivalent form: T30

```python
read = pd.read_csv(OUT / "read.csv", keep_default_na=False)
read[read["class"].str.startswith("same target")][["unit", "file", "page", "class", "quote", "conversion", "comment"]]
```

The 2023-Q2 report expects the cumulative integration-related expenses to be broadly offset by accretion effects of approximately USD 12bn. This is within 10% of the call figure of around 13bn, but indirect, published before the call figure and already recorded in phase 3b. T30 is counted separately from the call-only figures.

## Forward-looking sentences on a neighbouring quantity

```python
fw = read[(read.forward_looking == "yes") & (read["class"] == "other")]
fw.drop_duplicates(["unit", "comment"])[["unit", "file", "page", "quote", "comment"]]
```

These sentences are forward-looking but concern another period, another scope or no figure: the NCL targets for the end of 2025 and 2026 (T27 has its deadline at the end of 2024), the NCL key priority without a figure (the class B counterpart of phase 3d), and the quarterly amounts of integration-related expenses.

## Hits per form in pass A

```python
hits = pd.read_csv(OUT / "hits.csv", keep_default_na=False)
a = hits[hits.search_pass == "A"]
a.groupby(["unit", "form"]).size().rename("hits").reset_index()
```

Most pass A hits come from small percentages and amounts that occur near a keyword by chance, and from reported values: the going concern and leverage ratios, the currency sensitivity of the LRD (around USD 100bn for a 10% move of the US dollar) and the LRD level of USD 1,599.6bn at 31 March 2024, which is a result.

## Result per unit

```python
status = {"T33": "not mentioned", "T68_r1": "not mentioned", "T68_r2": "not mentioned", "F11": "not mentioned", "F22": "not mentioned", "T16": "mentioned without a figure", "T25": "mentioned without a figure", "T27": "mentioned without a figure", "F12": "mentioned without a figure", "T30": "counted separately"}
res = summary[["unit", "same target, same unit", "same target, another unit", "same target, another figure"]].copy()
res["status before"] = res.unit.map(status)
res["equivalent form found"] = (res.iloc[:, 1:4].sum(axis=1) > 0).map({True: "yes", False: "no"})
res
```

For the eight units T33, T68, F11, F22, T16, T25, T27 and F12, no report up to the end of 2024 states the target of the call in another unit or form. The count of call-only figures stays at 8, with T30 separate; the decision on T30 is Roman's.

## Limits

The forms are those listed in the plan; pass B guards against others but needs a keyword and a forward-looking word within 250 characters, so a target in a bare list would be missed. Not found means not found with these terms. The investor presentations were not searched.
